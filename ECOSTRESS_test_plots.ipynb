```python
def plot_eco_emissivity(file_path):
    with rasterio.open(file_path) as src:
        # Read the first band
        data = src.read(1).astype('float32')
        
        # 1. Apply Scaling Factor (ECOSTRESS V2 EmisWB is 0.0001)
        data *= 0.0001
        
        # 2. Mask NoData values (standard for ECOSTRESS is often 0 or -9999)
        # Using a small threshold to catch 0/negative raw values
        data[data <= 0] = np.nan 

        # 3. Create the plot
        fig, ax = plt.subplots(figsize=(10, 8))
        
        # Use 'viridis' or 'inferno' to highlight emissivity variations
        # Emissivity usually falls between 0.85 and 1.0
        img = ax.imshow(data, cmap='viridis', vmin=0.85, vmax=0.9)
        
        # Add cosmetic details
        plt.colorbar(img, ax=ax, label='Wideband Emissivity')
        ax.set_title(f"ECOSTRESS EmisWB\n{file_path.split('/')[-1]}")
        ax.set_axis_off() # Hide pixel coordinates for a cleaner look
        
        plt.show()

# Usage
# plot_eco_emissivity('your_cropped_file_name.tif')
```
